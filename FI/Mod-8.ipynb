## Monikko

Monikko on muuttumaton, järjestetty datarakenne.

## Joukko
Joukko on järjestelemätön datarakenne.
Joukon alkiot eivät siis ole tietyssä järjestyksessä, eikä niillä ole indeksejä.

Joukko voi sisältää tietyn arvon vain kerran.

Uusi joukko voidaan luoda seuraavasti:


```python
joukko1 = {1,2,3}
```

```python
joukko2 = set()
```

## Sisäkkäiset tietorakenteet

Tietorakenteiden alkioiden arvojen ei ole pakko olla alkeistyypin dataa kuten yksittäisiä numeroita tai merkkijonoja. 
Voimme myös tallentaa tietorakenteita toisten tietorakenteiden alkioiksi.

Voimme esimerkiksi luoda listan, jonka jokainen alkio on sanakirja seuraavasti:

```python

opiskelijat = [
    {
        "nimi": "Jane Doe",
        "Math": 2,
        "English": 3
    }, 
    {
            "nimi": "Sarah Smith",
            "Math": 5,
            "English": 1
    }, 
     {
                "nimi": "Sam Doe",
                "Math": 4,
                "English": 5
    }, 
]
```

Nyt voimme saada käsiimme jokaisen sanakirjan käyttämällä listan indeksejä. Esimerkiksi voimme tulostaa ensimmäisen listan alkion sisältämän sanakirjan:

```python
print(opiskelijat[0])
```

Tai käydä läpi kaikki listan alkiot vuoron perään:

```python
for opiskelija in opiskelijat:
    print(opiskelija)
```

Jos sitten haluamme käsiimme sanakirjan arvoja, haemme ensin listasta oikean sanakirjan listan indeksillä, ja sitten kyseisen sanakirjan sisältä avaimen arvon sen avaimen perusteella:

```python
opiskelijat[0]["nimi"]
```
