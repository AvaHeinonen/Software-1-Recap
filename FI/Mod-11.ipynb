## Yliluokkien ja aliluokkien määrittely

Voimme luoda yliluokan, ja aliluokan joka perii sen ominaisuudet seuraavasti:

```python
class Potion:
    def __init__(self, nimi):
        self.nimi = nimi

class Parannusjuoma(Potion):
    pass
```

Ylläoleva koodi luo luokat Potion ja Parannusjuoma.
Parannusjuoma on potionin aliluokka, ja perii kaikki sen ominaisuudet ja metodit

### Aliluokan alustaja ja aliluokan ominaisuuksien määrittely

Voimme määritellä aliluokalle sen omia, yliluokasta poikkeavia, ominaisuuksia luomalla sille oman alustajan. 

```python
class Potion:
    def __init__(self, nimi):
        self.nimi = nimi

class Parannusjuoma(Potion):
    def __init__(self, nimi, hp):
        super().__init__(nimi)
        self.hp = hp

juoma = Parannusjuoma("Terveysjuoma",2)
print(f"{juoma.nimi}: + {juoma.hp}")
```

## Metodien ylikirjoittaminen

Aliluokka voi *ylikirjoittaa* yliluokan metodin, eli aliluokalle voidaan määritellä oma toteutuksensa jostakin yliluokan metodista.

Voimme esimerkiksi määritellä Potion luokkaamme metodin käytä, joka vain tulostaa viestin komentoriville.
Voimme sitten ylikirjoittaa tämän metodin parannusjuomaluokassa siten, että se tulostaa viestin komentoriville, ja palauttaa tiedon siitä kuinka monta pistettä juomaa juomalla on parantunut.

Voimme jälleen kutsua yliluokan metodeja aliluokasta super() funktiota käyttämällä

```python
class Potion:
    def __init__(self, nimi):
        self.nimi = nimi

    def kayta(self):
        print(f"Käytät potionin")

class Parannusjuoma(Potion):
    def __init__(self, nimi, hp):
        super().__init__(nimi)
        self.hp = hp

    def kayta(self):
        super().kayta()
        return self.hp

juoma = Parannusjuoma("Terveysjuoma",2)
parannus = juoma.kayta()
print(f"Parannut {parannus} pistettä ")
```

Allaolevassa koodissa on määritelty luokat kenttä ja ajastettukenttä, joka perii luokan kenttä. 

Kentällä on maksimipisteet, sekä nykyiset pisteet.
Kentällä on metodit pelaa ja arvaa numero.

arvaa numero metodi pyytää käyttäjää arvaamaan numeroita, kunnes hän arvaa tietokoneen generoiman randomin kokonaisluvun oikein.

Pelaa metodi kutsuu arvaa numero metodia toistuvasti, kunnes käyttäjä on saanut vaaditun määrän pisteitä.


Muokkaa koodia siten että:
- Luokan ajastettu kenttä alustaja saa parametreikseen maksimipisteet ja aikarajan. se kutsuu yliluokkansa alustajaa, ja määrittelee sitten omat piirteensä aikaraja. 

- Luokka ajastettu kenttä ylikirjoittaa pelaa metodin siten että:
-- Se tallentaa pelin aloitusajan (koodissa on valmiiksi aloitusajan tallentaminen käyttäen time.time() metodia)
-- Se kutsuu toistuvasti arvaa_numero metodia niin kauan, kun pelaajalla ei ole riittävästi pisteitä. 
-- Joka kierroksella se laskee kuluneen ajan (aloitusaika on tallennettu muuttujaan aiemmin, voit käyttää sitä hyödyksesi), ja katsoo onko aikaa kulunut yli sallitun.
--- Jos aikaa on kulunut yli aikarajan, peli loppuu

```python
import random
import time
class Kenttä:
    def __init__(self, max_points):
        self.max_points = max_points
        self.current_points = 0
    def pelaa(self):
        while self.current_points != self.max_points:
            print(f"Vaaditut pisteet {self.max_points}")
            print(f"Nykyiset pisteet {self.current_points}")
            self.arvaa_numero()
    def arvaa_numero(self):
        numero = random.randint(1,5)
        arvaus = int(input("Arvaus: "))
        while arvaus != numero:
            if arvaus > numero:
                print("Liian korkea")
            else:
                print("Liian matala")
            
            arvaus = int(input("Arvaus: "))
        else:
            self.current_points += 1
            print("Oikein! 1 piste")

class AjastettuKenttä(Kenttä):
    def __init__(self, max_points, aikaraja):
        super().__init__(max_points)
        self.aikaraja = aikaraja


    def pelaa(self):
        aloitusaika = time.time()
        while self.current_points != self.max_points:
            print(f"Vaaditut pisteet {self.max_points}")
            print(f"Nykyiset pisteet {self.current_points}")
            pelin_kesto = time.time() - aloitusaika
            if pelin_kesto >= self.aikaraja:
                print(f"Aika loppui")
                break
            super().arvaa_numero()
       

kentta = AjastettuKenttä(3, 15)
kentta.pelaa()
```

## Moniperintä

Yhdellä aliluokalla voi olla useita yliluokkia.
Tätä kutsutaan *moniperinnäksi*

Alla luokka kala perii sekä luokat lemmikki että luokan eläin. 


```python
class Eläin:
    def __init__(self, laji):
        self.nimi = laji

class Lemmikki:
    def __init__(self, nimi, omistaja):
        self.nimi = nimi
        self.omistaja = omistaja

class Kala(Eläin, Lemmikki):
    pass
```

Nyt kun yliluokkia on useita, emme voi enää käyttää super() funktiota viitataksemme niihin, vaan meidän tulee viitata yliluokiin niiden nimillä.
Tehdäksemme alustajan luokalle Kala, voimme siis kutsua molempien yliluokkien alustajia seuraavasti:

```python
class Eläin:
    def __init__(self, laji):
        self.laji = laji

class Lemmikki:
    def __init__(self, nimi, omistaja):
        self.nimi = nimi
        self.omistaja = omistaja

class Kala(Eläin, Lemmikki):
    def __init__(self, nimi, omistaja, laji, vesistö):
        Eläin.__init__(self, laji)
        Lemmikki.__init__(self, nimi, omistaja)
        self.vesistö = vesistö

kala1 = Kala("Max", "Anna Armas", "Lohi", "Järvi")
print(f"{kala1.nimi}: {kala1.laji}, {kala1.vesistö}")
print(f"Omistaja: {kala1.omistaja}")
```
