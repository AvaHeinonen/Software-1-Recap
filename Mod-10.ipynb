# Assosiaatiot

Ohjelma koostuu useista luokista, joiden oliot toimivat vuorovaikutuksessa keskenään.
Assosiaatioilla viitataan ohjelman olioiden välisiin vuorovaikutussuhteisiin.

Usein tämä tarkoittaa sitä, että luokan ominaisuus on yksi tai useampi toisen luokan olio. 
Esimerkiksi suhde "Pelaajalla on lista esineitä jotka se on poiminut mukaansa" voidaan ohjelmassa luoda liittämällä pelaajan ominaisuudeksi lista "esineet" joka sisältää esine luokan olioita:

```python
class Pelaaja:
    def __init__(self, nimi, taso, esineet):
        self.nimi = nimi
        self.taso = taso
        self.esineet = esineet

class Esine:
    def __init__(self, nimi, hinta):
        self.nimi = nimi
        self.hinta = hinta

esineet = [Esine("Miekka", 123), Esine("Kilpi", 100)]
pelaaja1 = Pelaaja("Arthur", 0, esineet)
print(pelaaja1.esineet)
```

### Pysyvät ja tilapäiset suhteet
Ylempänä käsitellyt suhteet olioiden välillä ovat pysyviä. 
Yhden luokan olioon tallennetaan ominaisuutena toisen luokan olioita, joten kyseisten olioiden suhde kestää koko olioiden eliniän. 

Suhteet voivat myös olla väliaikaisia. 
Tällöin oliolle annetaan metodin parametrinä toisen luokan olio, ja luokka voi käyttää sen olion piirteitä ja metodeja oman metodinsa suorituksen ajan. 

Esimerkiksi allaolevassa Kenttä luokassa kenttä luokka ei tallenna monsteri luokan oliota, mutta sen metodi "printtaa monsteri" ottaa parametrina monsteriluokan olion, ja printtaa sen kuvauksen komentoriville

```python
class Monsteri:
    def __init__(self, nimi, kuvaus):
        self.nimi = nimi
        self.kuvaus = kuvaus

class Kentta:
    def __init__(self, taso, nimi, kuvaus):
        self.taso = taso
        self.nimi = nimi
        self.kuvaus = kuvaus
    def printtaa_monsteri(self, monsteri):
        print(f"{monsteri.nimi}: {monsteri.kuvaus}")

monsu1 = Monsteri("Kummitus", "Hurja kummitus")
kentta1 = Kentta(1, "Metsäpolku", "Helppo kävely nätin metsän läpi")
kentta1.printtaa_monsteri(monsu1)
```

### Isompi kertaustehtävä
Luo luokat huone, peli ja pelaaja.
Huoneen ominaisuuksia ovat nimi, kuvaus, ja esineet.
Vihjeet on lista merkkijonoja. 

huoneella on metodi etsi.
etsi metodi tulostaa komentoriville listan huoneessa olevista esineistä, ja pyytää käyttäjää valitsemaan esineen, jonka haluaa tallentaa esinelistaansa. 
Käyttäjä syöttää komentoriville sen esineen numeron, jonka haluaa tallentaa esinelistaansa. 
Metodi palauttaa pelaajan valitsemaa esinettä kuvaavan merkkijonon. 


Pelaajan ominaisuuksia ovat nimi ja esineet.
Esineet on lista merkkijonoja, ja se on aluksi tyhjä. 
Pelaajalla on myös metodi lisaa_esine, joka ottaa parametrina merkkijonon (esineen), ja tallentaa sen pelaajan esineet listaan.

Pelillä on ominaisuuksina huone ja pelaaja.
Pelillä on myös metodi etsi_huone, joka kutsuu huoneen etsi metodia, ja tallentaa sen palauttaman merkkijonon pelaajan esineet listaan käyttäen pelaajan lisaa_esine metodia. 

```python
class Huone:
    def __init__(self, nimi, kuvaus, esineet):
        self.nimi = nimi
        self.kuvaus = kuvaus
        self.esineet = esineet

    def etsi(self):
        for i in range(len(self.esineet)):
            print(f"{i}: {self.esineet[i]}")
        selection = int(input("Syötä valitsemasi esineen numero: "))
        return self.esineet[selection]

class Pelaaja:
    def __init__(self, nimi):
        self.nimi = nimi
        self.esineet = []

    def lisaa_esine(self, esine):
        self.esineet.append(esine)
        print(f"Olet lisännyt esineen {esine}")
        

class Peli:
    def __init__(self, pelaaja, huone):
        self.pelaaja = pelaaja
        self.huone = huone

    def etsi_huone(self):
        esine = self.huone.etsi()
        self.pelaaja.lisaa_esine(esine)

esineet = ["Miekka", "Kilpi", "Reppu", "Taikasauva"]
huone1 = Huone("Kellari", "Synkkä kellari", esineet)
pelaaja1 = Pelaaja("Sam")
peli1 = Peli(pelaaja1, huone1)
peli1.etsi_huone()

```
