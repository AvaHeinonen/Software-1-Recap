# Luokka, Olio, Alustaja
### Luokka ja luokan määrittely

Olio-ohjelmoinnissa **luokka** määrittelee jonkin ohjelman konseptin ominaisuudet ja toiminnot. 

**Olio** taas on luokan ajonaikainen instanssi, eli luokan kuvaaman konseptin ilmentymä, jolla on luokassa määritellyt ominaisuudet ja toiminnot.

Voimme esimerkiksi määritellä luokan pelaaja, jolla on ominaisuudet nimi, pelaaja id ja taso:

```python
class Pelaaja:
    def __init__(self, nimi, taso, pelaaja_ID):
        self.nimi = nimi
        self.taso = taso
        self.pelaaja_ID = pelaaja_ID
```

voimme sitten luoda uusia pelaaja-olioita, eli pelaajaluokan ajonaikaisia ilmentymiä seuraavasti:

```python
pelaaja1 = Pelaaja("Sandra", 0, 123)
pelaaja2 = Pelaaja("Kim", 1, 456)
```

### Luokkien metodit
Luokkien metodit ovat luokan sisällä määriteltyjä funktioita, joita voidaan kutsua luokan olioiden kautta.

Voimme lisätä pelaaja luokkaamme metodin level_up, joka päivittää kyseisen pelaajaolion tason:

```python
class Pelaaja:
    def __init__(self, nimi, taso, pelaaja_ID):
        self.nimi = nimi
        self.taso = taso
        self.pelaaja_ID = pelaaja_ID
    def level_up(self):
        self.taso += 1
```

Myös luokkien metodeihin annetaan aina ensimmäisenä parametrina self. 
Self viittaa siihen luokan olioon, joka metodia on kutsunut. 


Voimme kutsua luokan metodeita olioiden kautta:

```python
pelaaja3 = Pelaaja("Johan", 0, 123)
pelaaja3.level_up()
print(pelaaja3.taso)
```
