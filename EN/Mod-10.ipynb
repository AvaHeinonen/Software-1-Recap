# Associations

A program consists of multiple classes, whose objects interact with one another.
Associations refer to the interaction relationships between a program's objects.

Often this means that a class's property is one or more objects of another class. 
For example, the relationship "A player has a list of items they've picked up" can be implemented in a program by giving the player an "items" property — a list that holds objects of the Item class:

```python
class Player:
    def __init__(self, name, level, items):
        self.name = name
        self.level = level
        self.items = items

class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

items = [Item("Sword", 123), Item("Shield", 100)]
player1 = Player("Arthur", 0, items)
print(player1.items)
```

### Direction of Relationships

Relationships like the one above are one-directional. 
The player knows about both items and rooms, but an item or a room doesn't know about the player.

A relationship can also be two-directional. In that case, both classes know about each other.
For example, we can build the classes Item and Room so that the room has a list of the items in it, and the item has the room it was added to. 
Now the room's item list starts out empty, and items are added to it later in the main program.


```python
class Room:
    def __init__(self, name, description):
        self.name = name
        self.description = description
        self.items = []

    def add_item(self, item):
        self.items.append(item)

class Item:
    def __init__(self, name, price, room):
        self.name = name
        self.price = price
        self.room = room

room1 = Room("Hallway", "A messy hallway")
item1 = Item("sword", 150, room1)
room1.add_item(item1)

print(room1.name)
print(item1.room.name)
print(room1.items[0].name)
print(item1.name)
```

### Permanent and Temporary Relationships
The relationships between objects covered above are permanent. 
An object of one class stores objects of another class as a property, so the relationship between those objects lasts for the entire lifetime of the objects. 

Relationships can also be temporary. 
In that case, an object of another class is given to an object as a method parameter, and the class can use that object's properties and methods for the duration of its own method's execution. 

For example, in the Field class below, the Field class doesn't store the Monster class object, but its method "print_monster" takes a Monster class object as a parameter, and prints its description to the command line

```python
class Monster:
    def __init__(self, name, description):
        self.name = name
        self.description = description

class Field:
    def __init__(self, level, name, description):
        self.level = level
        self.name = name
        self.description = description
    def print_monster(self, monster):
        print(f"{monster.name}: {monster.description}")

monster1 = Monster("Ghost", "A fierce ghost")
field1 = Field(1, "Forest Path", "An easy walk through a nice forest")
field1.print_monster(monster1)
```
