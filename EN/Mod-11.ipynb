# Inheritance

In Python we can create a class hierarchy, with *superclasses* and *subclasses*.
Subclasses have all the properties of the superclass, but they can also have their own properties and functionality.

## Defining Superclasses and Subclasses

We can create a superclass, and a subclass that inherits its properties, as follows:

```python
class Potion:
    def __init__(self, name):
        self.name = name

class HealthPotion(Potion):
    pass
```

The code above creates the classes Potion and HealthPotion.
HealthPotion is a subclass of Potion, and inherits all of its properties and methods

### Subclass Initializer and Defining Subclass Properties

We can define properties specific to a subclass, different from the superclass, by creating an initializer for it. 

```python
class Potion:
    def __init__(self, name):
        self.name = name

class HealthPotion(Potion):
    def __init__(self, name, hp):
        super().__init__(name)
        self.hp = hp

potion = HealthPotion("Health Potion",2)
print(f"{potion.name}: + {potion.hp}")
```

## Overriding Methods

A subclass can *override* a superclass method, meaning a subclass can be given its own implementation of a method from the superclass.

For example, we can define a method use in our Potion class, which just prints a message to the command line.
We can then override this method in the HealthPotion class, so that it prints a message to the command line, and returns how many points of health the potion restored when used.

We can again call the superclass's methods from the subclass using the super() function

```python
class Potion:
    def __init__(self, name):
        self.name = name

    def use(self):
        print(f"You use the potion")

class HealthPotion(Potion):
    def __init__(self, name, hp):
        super().__init__(name)
        self.hp = hp

    def use(self):
        super().use()
        return self.hp

potion = HealthPotion("Health Potion",2)
healing = potion.use()
print(f"You are healed {healing} points ")
```

## Multiple Inheritance

A single subclass can have multiple superclasses.
This is called *multiple inheritance*

Below, the class Fish inherits both the Pet and the Animal classes. 


```python
class Animal:
    def __init__(self, species):
        self.name = species

class Pet:
    def __init__(self, name, owner):
        self.name = name
        self.owner = owner

class Fish(Animal, Pet):
    pass
```

Now that there are multiple superclasses, we can no longer use the super() function to refer to them, and instead need to refer to the superclasses by name.
To create an initializer for the Fish class, we can therefore call both superclasses' initializers as follows:

```python
class Animal:
    def __init__(self, species):
        self.species = species

class Pet:
    def __init__(self, name, owner):
        self.name = name
        self.owner = owner

class Fish(Animal, Pet):
    def __init__(self, name, owner, species, body_of_water):
        Animal.__init__(self, species)
        Pet.__init__(self, name, owner)
        self.body_of_water = body_of_water

fish1 = Fish("Max", "Anna Armas", "Salmon", "Lake")
print(f"{fish1.name}: {fish1.species}, {fish1.body_of_water}")
print(f"Owner: {fish1.owner}")
```
