## Tuple

A tuple is an immutable, ordered data structure.
**Immutable**: You cannot add elements to a tuple, or remove elements from it, once it has been created
**Ordered**: The elements are stored in a specific order, and each element has an *index*, i.e. a position number, that tells its place within the tuple



### Creating a Tuple and Reading its Elements
A tuple can be created as follows:

```python
tuple1 = (1,2,3)
```

The parentheses aren't required, though — the following also works:

```python
tuple2 = 1,2,3
```

Since a tuple is ordered, each of its elements has an index. Using the index, we can access the value of a single element in the tuple, or the values of several consecutive elements.
The syntax for this is the same as when working with lists.

```python
# Create a new tuple
tuple3 = 1,2,3,4,5,6
# Print the value of the tuple's first element
print(tuple3[0])
# print the values of the tuple's first three elements
print(tuple3[0:3])
```

When we want to return multiple values from a function, we can store the return values in a tuple, and then unpack them into variables when calling the function, as follows

```python
def do_math(number1, number2):
    total = number1 + number2
    product = number1 * number2
    return (total, product)

(total, product) = do_math(1,2)

print(total)
print(product)
```

## Set
A set is an unordered data structure.
So the elements of a set aren't in any particular order, and they don't have indices.

A set can contain a given value only once.

### Creating a Set
A new set can be created as follows:


```python
set1 = {1,2,3}
```

Unlike, for example, lists, we can't create a new empty set with the syntax set2 = {}.
Instead, a new, empty set is created as follows:

```python
set2 = set()
```

### Adding and Removing Elements
Elements can be added to, and removed from, a set while the program is running. 

New elements can be added using the add method:

```python
set2.add("Matt")
print(set2)
```

Elements of a set can be removed based on their value as follows:

```python
set2.remove("Matt")
print(set2)
```

## Nested Data Structures

The elements of a data structure don't have to hold primitive data such as single numbers or strings. 
We can also store data structures as elements of other data structures.

For example, we can create a list in which each element is a dictionary, as follows:

```python

students = [
    {
        "name": "Jane Doe",
        "Math": 2,
        "English": 3
    }, 
    {
            "name": "Sarah Smith",
            "Math": 5,
            "English": 1
    }, 
     {
                "name": "Sam Doe",
                "Math": 4,
                "English": 5
    }, 
]
```

Now we can access each dictionary using the list's indices. For example, we can print the dictionary contained in the first element of the list:

```python
print(students[0])
```

Or go through all the elements of the list one at a time:

```python
for student in students:
    print(student)
```

If we then want to get a value out of a dictionary, we first fetch the correct dictionary from the list using the list index, and then get the value inside that dictionary based on its key:

```python
students[0]["name"]
```
