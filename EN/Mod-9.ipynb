# Class, Object, Initializer
### Class and Class Definition

In object-oriented programming, a **class** defines the properties and behaviors of some program concept. 

An **object**, on the other hand, is a runtime instance of a class — that is, an instantiation of the concept the class describes, with the properties and behaviors defined in the class.

For example, we can define a class Player, which has the properties name, player ID and level:

```python
class Player:
    def __init__(self, name, level, player_id):
        self.name = name
        self.level = level
        self.player_id = player_id
```

we can then create new player objects, i.e. runtime instances of the Player class, as follows:

```python
player1 = Player("Sandra", 0, 123)
player2 = Player("Kim", 1, 456)
```

### Initializer

### Class Methods
Class methods are functions defined inside a class, which can be called through the class's objects.

We can add the method level_up to our Player class, which updates that player object's level:

```python
class Player:
    def __init__(self, name, level, player_id):
        self.name = name
        self.level = level
        self.player_id = player_id
    def level_up(self):
        self.level += 1
```

Class methods are also always given self as their first parameter. 
Self refers to the class object that called the method. 


We can call a class's methods through its objects:

```python
player3 = Player("Johan", 0, 123)
player3.level_up()
print(player3.level)
```
